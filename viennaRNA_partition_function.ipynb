```python
import RNA

sequence = "GGGAAACCC"
target_structure = "(((...)))"

fc = RNA.fold_compound(sequence)
mfe_struct, mfe_energy = fc.mfe()
fc.pf()
bpp = fc.bpp()

if target_structure is None:
    target_structure = mfe_struct

target_pairs = []
stack = []
for i, char in enumerate(target_structure):
    if char == '(':
        stack.append(i + 1)
    elif char == ')':
        target_pairs.append((stack.pop(), i + 1))

target_pair_set = set(target_pairs)
defect = 0.0
length = len(sequence)

for i, j in target_pairs:
    defect += (1 - bpp[i][j])

for i in range(1, length + 1):
    for j in range(i + 4, length + 1):
        if (i, j) not in target_pair_set:
            defect += bpp[i][j]

print(f"{sequence} | {target_structure} | ED: {defect:.3f}")
```

```python
def calculate_pf(sequence):
    fc = RNA.fold_compound(sequence)
    pf_result = fc.pf()
    Z = pf_result[1]  # Partition function
    bpp_matrix = fc.bpp()

    return {
        'partition_function': Z,
        'base_pair_probabilities': bpp_matrix,
        'sequence': sequence
    }

sequence = "GGGAAACCC"
result = calculate_pf(sequence)

print(f"Pf Z: {result['partition_function']:.6f}")
print("\nBase pair probabilities p(i,j):")

bpp = result['base_pair_probabilities']
length = len(sequence)

for i in range(1, length + 1):
    for j in range(i + 1, length + 1):
        prob = bpp[i][j]
        print(f"p({i},{j}) = {prob:.3f}")
```
